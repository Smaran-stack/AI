```python
import numpy as np


empty_array = np.empty((2, 2))
print("Empty Array:\n", empty_array)


ones_array = np.ones((4, 2))
print("\nAll-One Array:\n", ones_array)

filled_array = np.full((3, 3), 7)
print("\nFilled Array:\n", filled_array)

zeros_like_array = np.zeros_like(filled_array)
print("\nZeros Like Array:\n", zeros_like_array)


ones_like_array = np.ones_like(filled_array)
print("\nOnes Like Array:\n", ones_like_array)


new_list = [1, 2, 3, 4]
array_from_list = np.array(new_list)
print("\nArray from list:\n", array_from_list)
```

```python
arr_range = np.arange(10, 50)
print("Array from 10 to 49:\n", arr_range)

matrix_3x3 = np.arange(9).reshape((3, 3))
print("\n3x3 Matrix:\n", matrix_3x3)

identity_3x3 = np.eye(3)
print("\nIdentity Matrix:\n", identity_3x3)

rand_arr = np.random.random(30)
print("\nRandom Array:\n", rand_arr)
print("Mean:", rand_arr.mean())

rand_10x10 = np.random.random((10, 10))
print("\nMin:", rand_10x10.min(), "Max:", rand_10x10.max())

zero_array = np.zeros(10)
zero_array[4] = 1
print("\nModified Zero Array:\n", zero_array)

arr = np.array([1, 2, 0, 0, 4, 0])
reversed_arr = arr[::-1]
print("\nReversed Array:\n", reversed_arr)

# 8. 2D array with 1 on border and 0 inside
border_array = np.ones((5, 5))
border_array[1:-1, 1:-1] = 0
print("\n2D Array with border 1s:\n", border_array)

checkerboard = np.zeros((8, 8))
checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1
print("\nCheckerboard Pattern:\n", checkerboard)
```

```python
x = np.array([[1, 2], [3, 5]])
y = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
w = np.array([11, 12])

print("x + y:\n", x + y)

print("x - y:\n", x - y)

print("x * 3:\n", x * 3)

print("x^2:\n", x**2)

print("v.w:", np.dot(v, w))
print("x.v:", np.dot(x, v))
print("x.y:\n", np.dot(x, y))

print("x and y along row:\n", np.concatenate((x, y), axis=0))
print("v and w along column:\n", np.vstack((v, w)))
```

```python
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])

# 1. A * A^-1 = I
A_inv = np.linalg.inv(A)
print("A * A^-1:\n", np.dot(A, A_inv))

# 2. AB != BA
print("AB:\n", np.dot(A, B))
print("BA:\n", np.dot(B, A))

# 3. (AB)^T = B^T A^T
print("Transpose check:\n", np.dot(A, B).T)
print("B^T A^T:\n", np.dot(B.T, A.T))

# Solve system of equations
# 2x - 3y + z = -1
# x - y + 2z = -3
# 3x + y - z = 9

coeff = np.array([[2, -3, 1], [1, -1, 2], [3, 1, -1]])
const = np.array([-1, -3, 9])
solution = np.linalg.inv(coeff).dot(const)
print("Solution:\n", solution)
```

```python
import time

size = 1000000

# Python lists
list1 = list(range(size))
list2 = list(range(size))

# Numpy arrays
np1 = np.array(list1)
np2 = np.array(list2)

# 1. Element-wise addition
start = time.time()
py_add = [a + b for a, b in zip(list1, list2)]
print("Python list addition time:", time.time() - start)

start = time.time()
np_add = np1 + np2
print("Numpy addition time:", time.time() - start)

# 2. Element-wise multiplication
start = time.time()
py_mul = [a * b for a, b in zip(list1, list2)]
print("Python list multiplication time:", time.time() - start)

start = time.time()
np_mul = np1 * np2
print("Numpy multiplication time:", time.time() - start)

# 3. Dot product
start = time.time()
py_dot = sum(a*b for a, b in zip(list1, list2))
print("Python list dot product time:", time.time() - start)

start = time.time()
np_dot = np.dot(np1, np2)
print("Numpy dot product time:", time.time() - start)

# 4. Matrix multiplication 1000x1000
size_mat = 1000
py_mat1 = [list(range(size_mat)) for _ in range(size_mat)]
py_mat2 = [list(range(size_mat)) for _ in range(size_mat)]
np_mat1 = np.array(py_mat1)
np_mat2 = np.array(py_mat2)

start = time.time()
# Using pure Python nested loops (very slow!)
py_result = [[sum(a*b for a,b in zip(row,col)) for col in zip(*py_mat2)] for row in py_mat1]
print("Python list matrix multiplication time:", time.time() - start)

start = time.time()
np_result = np.dot(np_mat1, np_mat2)
print("Numpy matrix multiplication time:", time.time() - start)
```
